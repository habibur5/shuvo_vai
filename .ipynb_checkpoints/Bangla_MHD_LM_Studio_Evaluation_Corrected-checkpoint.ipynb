## Dataset loading correction

**Do not change `header=None` for this file.**

Your actual CSV is effectively:

```text
Bangla text...,0
Bangla text...,1
Bangla text...,0
...
```

There is no `text,label` header row.

````python
# ============================================================
# LOAD DATASET
# ============================================================

if not Path(DATA_PATH).exists():
    raise FileNotFoundError(
        f"Could not find {DATA_PATH}. "
        "Put it beside the notebook or change DATA_PATH."
    )

raw_df = pd.read_csv(
    DATA_PATH,
    header=None,
    names=["text", "label"],
    encoding="utf-8",
    dtype=str,
)

print("Raw shape:", raw_df.shape)
display(raw_df.head())
````

````python
# ============================================================
# CLEAN DATASET
# ============================================================

def clean_text(value):
    if pd.isna(value):
        return ""

    value = unicodedata.normalize("NFKC", str(value))
    value = value.replace("\ufeff", "").replace("\u00a0", " ")
    value = re.sub(r"\s+", " ", value)
    return value.strip()


df = raw_df.copy()
df["text"] = df["text"].map(clean_text)

df["label"] = pd.to_numeric(
    df["label"].astype(str).str.strip(),
    errors="coerce",
)

print("Raw rows:", len(df))
print("Empty text:", df["text"].eq("").sum())
print("Missing labels:", df["label"].isna().sum())
print("Invalid labels:", (~df["label"].isin([0, 1])).sum())

df = df[
    (df["text"] != "") &
    (df["label"].isin([0, 1]))
].copy()

df["label"] = df["label"].astype(int)

before_dup = len(df)
df = df.drop_duplicates(subset=["text"]).reset_index(drop=True)
print("Duplicates removed:", before_dup - len(df))

df.insert(0, "row_id", np.arange(len(df)))

print("\nFINAL DATASET:", df.shape)
print(df["label"].value_counts().sort_index())

display(
    df.sample(10, random_state=SEED)[["row_id", "text", "label"]]
)
````

````python
# ============================================================
# DATASET HASH + SANITY CHECK
# ============================================================

def dataframe_hash(frame):
    payload = frame[
        ["row_id", "text", "label"]
    ].to_csv(index=False).encode("utf-8")
    return hashlib.sha256(payload).hexdigest()

DATASET_HASH = dataframe_hash(df)

assert len(df) == 7109, (
    f"Expected 7109 usable rows for the uploaded dataset, "
    f"but got {len(df)}. Check whether Dataset(2).csv changed."
)

assert df["label"].value_counts().to_dict() == {0: 3511, 1: 3598}

print("Dataset check PASSED.")
print("SHA-256:", DATASET_HASH)
````

## Few-shot demonstrations

These are the **same manual prefixes from your uploaded code**. The notebook checks that each exists in the current dataset and has the expected label.

The demonstration rows are removed before DEV/TEST creation, so they cannot leak into evaluation.

````python
# ============================================================
# MANUAL DEMOS
# ============================================================

POS_PREFIXES = [
    "রাতে ঘুম হয় না টেনশনে",
    "আমি আসলে মরে যেতে চাইনা। কিন্তু বেঁচে থাকার",
    "আমি বার বার একটা শব্দ বা নাম উচ্চারণ করতে থাকি",
    "নিজের ফেমেলি প্রবলেম গুলো কাওকে বলতেও পারি না",
    "আমার এংজাইটি ডিসঅর্ডার হয় প্রায় দেড় বছর আগে",
]

NEG_PREFIXES = [
    "যে কোন রিলেশন ব্রেকআপের কারনে সুইসাইড করতে ইচ্ছে হলে",
    "আমি মুগ্ধ হয়ে আমার মেয়ের দিকে তাকিয়ে আছি",
    "#মায়াবতী #পর্ব:৩৬ #তানিশা সুলতানা",
    "মুখ ফুটে কখনো বলা হয়নি মা তোমায় অনেক ভালোবাসি",
    "🥰আলহামদুলিল্লাহ 🥰 অনেক জল্পনা কল্পনা শেষে সব বিরোধী দলের",
]


def find_demo(prefix, expected_label):
    prefix = clean_text(prefix)

    matches = df[
        df["text"].str.startswith(prefix, na=False)
    ]

    if len(matches) == 0:
        raise ValueError(f"Demo not found: {prefix}")

    row = matches.iloc[0]

    if int(row["label"]) != expected_label:
        raise ValueError(
            f"Demo label mismatch for: {prefix}\\n"
            f"Expected {expected_label}, found {row['label']}"
        )

    return row


pos_demo_rows = [find_demo(x, 1) for x in POS_PREFIXES]
neg_demo_rows = [find_demo(x, 0) for x in NEG_PREFIXES]

demo_rows = pd.DataFrame(
    pos_demo_rows + neg_demo_rows
).drop_duplicates("row_id")

print("MHD demos:", len(pos_demo_rows))
print("NoMHD demos:", len(neg_demo_rows))
print("Unique demo rows:", len(demo_rows))

display(
    demo_rows[["row_id", "label", "text"]]
)
````

````python
# ============================================================
# MESSAGE BUILDING
# ============================================================

def clip_text(text, max_chars):
    text = clean_text(text)
    if len(text) <= max_chars:
        return text
    return text[:max_chars].rstrip() + "…"


def build_messages(text, shots):
    messages = [
        {
            "role": "system",
            "content": SYSTEM_PROMPT,
        }
    ]

    if shots not in SHOTS_PER_CLASS:
        raise ValueError(f"Unsupported shots: {shots}")

    # Contrastive in-context examples:
    # MHD example -> 1
    # NoMHD example -> 0
    for i in range(shots):
        messages.append({
            "role": "user",
            "content": USER_TEMPLATE.format(
                text=clip_text(
                    pos_demo_rows[i]["text"],
                    MAX_SHOT_CHARS,
                )
            ),
        })
        messages.append({
            "role": "assistant",
            "content": "1",
        })

        messages.append({
            "role": "user",
            "content": USER_TEMPLATE.format(
                text=clip_text(
                    neg_demo_rows[i]["text"],
                    MAX_SHOT_CHARS,
                )
            ),
        })
        messages.append({
            "role": "assistant",
            "content": "0",
        })

    messages.append({
        "role": "user",
        "content": USER_TEMPLATE.format(
            text=clip_text(text, MAX_TEST_CHARS)
        ),
    })

    return messages


for k in SHOTS_PER_CLASS:
    print(k, "shot =>", len(build_messages("টেস্ট লেখা", k)), "messages")
````

````python
# ============================================================
# STRICT PARSER + PREDICTION
# ============================================================

def parse_prediction(response_text):
    if response_text is None:
        return None

    text = str(response_text).strip()

    if text == "0":
        return 0
    if text == "1":
        return 1

    cleaned = text.replace("```", "").strip()

    if re.fullmatch(r"উত্তর\s*[:：]?\s*0[।.]?", cleaned):
        return 0

    if re.fullmatch(r"উত্তর\s*[:：]?\s*1[।.]?", cleaned):
        return 1

    # Deliberately reject:
    # "The answer is 0"
    # "0 because..."
    # "0, not 1"
    return None


def predict_one(text, shots):
    messages = build_messages(text, shots)
    start = time.perf_counter()
    last_error = ""

    for attempt in range(1, RETRY_COUNT + 1):
        try:
            response = client.chat.completions.create(
                model=MODEL_NAME,
                messages=messages,
                temperature=TEMPERATURE,
                max_tokens=MAX_TOKENS,
                stream=False,
            )

            raw_output = (
                response.choices[0].message.content
                if response.choices
                else ""
            )

            prediction = parse_prediction(raw_output)

            return {
                "prediction": prediction,
                "raw_output": raw_output,
                "valid_format": prediction in [0, 1],
                "error": "",
                "latency_sec": time.perf_counter() - start,
                "attempts": attempt,
            }

        except Exception as e:
            last_error = repr(e)

            if attempt < RETRY_COUNT:
                time.sleep(
                    SLEEP_BETWEEN_RETRIES * (2 ** (attempt - 1))
                )

    return {
        "prediction": None,
        "raw_output": "",
        "valid_format": False,
        "error": last_error,
        "latency_sec": time.perf_counter() - start,
        "attempts": RETRY_COUNT,
    }


print("Prediction function ready.")
````

````python
# ============================================================
# TEST ONE REQUEST
# ============================================================

sample = df.iloc[0]

print("TEXT:")
print(sample["text"][:700])

result = predict_one(sample["text"], shots=0)

print("\nRAW MODEL OUTPUT:", repr(result["raw_output"]))
print("PARSED:", result["prediction"])
print("VALID FORMAT:", result["valid_format"])
print("LATENCY:", round(result["latency_sec"], 3), "sec")
````

````python
# ============================================================
# DEV / TEST SPLIT
# ============================================================

demo_ids = set(demo_rows["row_id"].astype(int))

remaining = df[
    ~df["row_id"].isin(demo_ids)
].copy()

dev_parts = []

for label in [0, 1]:
    part = remaining[remaining["label"] == label]

    dev_parts.append(
        part.sample(
            n=DEV_PER_CLASS,
            random_state=SEED + label,
        )
    )

dev_df = (
    pd.concat(dev_parts)
    .sample(frac=1, random_state=SEED)
    .reset_index(drop=True)
)

dev_ids = set(dev_df["row_id"].astype(int))

test_df = remaining[
    ~remaining["row_id"].isin(dev_ids)
].reset_index(drop=True)

print("Demo:", len(demo_rows))
print("DEV:", len(dev_df))
print("TEST:", len(test_df))

print("\nDEV:")
print(dev_df["label"].value_counts().sort_index())

print("\nTEST:")
print(test_df["label"].value_counts().sort_index())

dev_df.to_csv(
    OUT_DIR / "dev_set.csv",
    index=False,
    encoding="utf-8-sig",
)

test_df.to_csv(
    OUT_DIR / "test_set.csv",
    index=False,
    encoding="utf-8-sig",
)

demo_rows.to_csv(
    OUT_DIR / "demo_rows.csv",
    index=False,
    encoding="utf-8-sig",
)
````

````python
# ============================================================
# PILOT / FULL EVALUATION SET
# ============================================================

if RUN_PILOT:
    parts = []

    for label in [0, 1]:
        part = test_df[test_df["label"] == label]

        parts.append(
            part.sample(
                n=min(PILOT_N_PER_CLASS, len(part)),
                random_state=SEED + 100 + label,
            )
        )

    eval_df = (
        pd.concat(parts)
        .sample(frac=1, random_state=SEED)
        .reset_index(drop=True)
    )

    print("PILOT MODE:", len(eval_df), "rows")
else:
    eval_df = test_df.copy()
    print("FULL TEST MODE:", len(eval_df), "rows")

print(eval_df["label"].value_counts().sort_index())
````

````python
# ============================================================
# EXPERIMENT CONFIGURATION
# ============================================================

def dataframe_hash(frame):
    payload = frame[
        ["row_id", "text", "label"]
    ].to_csv(index=False).encode("utf-8")
    return hashlib.sha256(payload).hexdigest()


TEST_HASH = dataframe_hash(test_df)

config = {
    "created_at": datetime.now().isoformat(),
    "model_name": MODEL_NAME,
    "base_url": LM_STUDIO_BASE_URL,
    "dataset_path": DATA_PATH,
    "dataset_hash": DATASET_HASH,
    "test_hash": TEST_HASH,
    "prompt_version": PROMPT_VERSION,
    "shots_per_class": SHOTS_PER_CLASS,
    "max_test_chars": MAX_TEST_CHARS,
    "max_shot_chars": MAX_SHOT_CHARS,
    "dev_per_class": DEV_PER_CLASS,
    "temperature": TEMPERATURE,
    "max_tokens": MAX_TOKENS,
    "seed": SEED,
    "pilot": RUN_PILOT,
    "pilot_n_per_class": PILOT_N_PER_CLASS,
}

with open(
    OUT_DIR / "experiment_config.json",
    "w",
    encoding="utf-8",
) as f:
    json.dump(config, f, ensure_ascii=False, indent=2)

with open(
    OUT_DIR / "prompt.txt",
    "w",
    encoding="utf-8",
) as f:
    f.write("SYSTEM PROMPT\n")
    f.write(SYSTEM_PROMPT)
    f.write("\n\nUSER TEMPLATE\n")
    f.write(USER_TEMPLATE)

print("Experiment configuration saved.")
````

````python
# ============================================================
# SAFE CHECKPOINTED RUNNER
# ============================================================

def setting_name(k):
    return "Zero-shot" if k == 0 else f"{k}-shot"


def validate_checkpoint(checkpoint, current_eval):
    required = {
        "row_id", "text", "label", "prediction",
        "raw_output", "valid_format", "error",
        "latency_sec", "attempts"
    }

    missing = required - set(checkpoint.columns)

    if missing:
        raise ValueError(
            f"Checkpoint missing columns: {sorted(missing)}"
        )

    expected = current_eval.set_index("row_id")

    for _, row in checkpoint.iterrows():
        rid = int(row["row_id"])

        if rid not in expected.index:
            raise ValueError(
                f"Checkpoint row_id={rid} is not in current evaluation set."
            )

        if str(row["text"]) != str(expected.loc[rid, "text"]):
            raise ValueError(
                f"Text mismatch for row_id={rid}. "
                "Do not resume this checkpoint with a different split."
            )

        if int(row["label"]) != int(expected.loc[rid, "label"]):
            raise ValueError(
                f"Label mismatch for row_id={rid}."
            )


def run_setting(k):
    name = setting_name(k)

    final_path = OUT_DIR / f"preds_{name}.csv"
    partial_path = OUT_DIR / f"partial_{name}.csv"

    print("\n" + "=" * 90)
    print(name)
    print("=" * 90)

    if final_path.exists():
        existing = pd.read_csv(
            final_path,
            encoding="utf-8-sig",
        )
        validate_checkpoint(existing, eval_df)
        print("[LOADED FINAL]", final_path)
        return existing

    completed = {}

    if partial_path.exists():
        partial = pd.read_csv(
            partial_path,
            encoding="utf-8-sig",
        )

        validate_checkpoint(partial, eval_df)

        for _, row in partial.iterrows():
            completed[int(row["row_id"])] = row.to_dict()

        print("[RESUME]", len(completed), "rows")

    results = []

    for _, row in tqdm(
        eval_df.iterrows(),
        total=len(eval_df),
        desc=name,
    ):
        rid = int(row["row_id"])

        if rid in completed:
            results.append(completed[rid])
            continue

        out = predict_one(
            text=row["text"],
            shots=k,
        )

        item = {
            "row_id": rid,
            "text": row["text"],
            "label": int(row["label"]),
            "prediction": out["prediction"],
            "raw_output": out["raw_output"],
            "valid_format": out["valid_format"],
            "error": out["error"],
            "latency_sec": out["latency_sec"],
            "attempts": out["attempts"],
        }

        completed[rid] = item
        results.append(item)

        if len(completed) % CHECKPOINT_EVERY == 0:
            checkpoint = pd.DataFrame(
                list(completed.values())
            )

            checkpoint.to_csv(
                partial_path,
                index=False,
                encoding="utf-8-sig",
            )

    final_df = pd.DataFrame(list(completed.values()))

    # Preserve evaluation order.
    final_df = (
        eval_df[["row_id", "text", "label"]]
        .merge(
            final_df.drop(
                columns=["text", "label"],
                errors="ignore",
            ),
            on="row_id",
            how="left",
        )
    )

    final_df.to_csv(
        final_path,
        index=False,
        encoding="utf-8-sig",
    )

    # Keep the same complete file as recovery checkpoint.
    final_df.to_csv(
        partial_path,
        index=False,
        encoding="utf-8-sig",
    )

    print("[SAVED]", final_path)

    return final_df
````

````python
# ============================================================
# METRICS
# ============================================================

def calculate_metrics(result_df):
    valid = result_df[
        result_df["prediction"].isin([0, 1])
    ].copy()

    n_total = len(result_df)
    n_valid = len(valid)
    n_invalid = n_total - n_valid

    if n_valid == 0:
        return {
            "N_total": n_total,
            "N_valid": 0,
            "N_invalid": n_invalid,
            "Format_Adherence": 0.0,
            "Accuracy": np.nan,
            "Macro_Precision": np.nan,
            "Macro_Recall": np.nan,
            "Macro_F1": np.nan,
            "Weighted_F1": np.nan,
            "MCC": np.nan,
        }

    y_true = valid["label"].astype(int)
    y_pred = valid["prediction"].astype(int)

    return {
        "N_total": n_total,
        "N_valid": n_valid,
        "N_invalid": n_invalid,
        "Format_Adherence": n_valid / n_total,
        "Accuracy": accuracy_score(y_true, y_pred),
        "Macro_Precision": precision_score(
            y_true, y_pred, average="macro", zero_division=0
        ),
        "Macro_Recall": recall_score(
            y_true, y_pred, average="macro", zero_division=0
        ),
        "Macro_F1": f1_score(
            y_true, y_pred, average="macro", zero_division=0
        ),
        "Weighted_F1": f1_score(
            y_true, y_pred, average="weighted", zero_division=0
        ),
        "MCC": matthews_corrcoef(y_true, y_pred),
    }


metric_rows = []

for k, result_df in preds.items():
    row = calculate_metrics(result_df)
    row["Setting"] = setting_name(k)
    row["Shots_per_class"] = k
    metric_rows.append(row)

metrics_df = pd.DataFrame(metric_rows)

metrics_df = metrics_df[
    [
        "Setting",
        "Shots_per_class",
        "N_total",
        "N_valid",
        "N_invalid",
        "Format_Adherence",
        "Accuracy",
        "Macro_Precision",
        "Macro_Recall",
        "Macro_F1",
        "Weighted_F1",
        "MCC",
    ]
]

display(metrics_df)

metrics_df.to_csv(
    OUT_DIR / "metrics_summary.csv",
    index=False,
    encoding="utf-8-sig",
)
````

````python
# ============================================================
# CLASSIFICATION REPORTS + CONFUSION MATRICES
# ============================================================

for k, result_df in preds.items():
    name = setting_name(k)

    valid = result_df[
        result_df["prediction"].isin([0, 1])
    ].copy()

    if len(valid) == 0:
        continue

    print("\n" + "=" * 80)
    print(name)
    print("=" * 80)

    print(
        classification_report(
            valid["label"].astype(int),
            valid["prediction"].astype(int),
            labels=[0, 1],
            target_names=["NoMHD", "MHD"],
            zero_division=0,
        )
    )

    cm = confusion_matrix(
        valid["label"].astype(int),
        valid["prediction"].astype(int),
        labels=[0, 1],
    )

    plt.figure(figsize=(5, 4))
    sns.heatmap(
        cm,
        annot=True,
        fmt="d",
        xticklabels=["NoMHD", "MHD"],
        yticklabels=["NoMHD", "MHD"],
    )
    plt.title(f"Confusion Matrix — {name}")
    plt.xlabel("Predicted")
    plt.ylabel("Actual")
    plt.tight_layout()

    plt.savefig(
        OUT_DIR / f"confusion_matrix_{name}.png",
        dpi=300,
        bbox_inches="tight",
    )

    plt.show()
    plt.close()
````

````python
# ============================================================
# OVERALL PERFORMANCE PLOT
# ============================================================

metric_cols = [
    "Accuracy",
    "Macro_Precision",
    "Macro_Recall",
    "Macro_F1",
    "Weighted_F1",
]

ax = metrics_df.set_index("Setting")[
    metric_cols
].plot(
    kind="bar",
    figsize=(10, 6),
)

ax.set_ylim(0, 1)
ax.set_ylabel("Score")
ax.set_title("Bangla MHD Classification Performance")
plt.xticks(rotation=0)
plt.tight_layout()

plt.savefig(
    OUT_DIR / "overall_metrics.png",
    dpi=300,
    bbox_inches="tight",
)

plt.show()
plt.close()
````

````python
# ============================================================
# ERROR + INVALID OUTPUT ANALYSIS
# ============================================================

all_errors = []
all_invalid = []

for k, result_df in preds.items():
    name = setting_name(k)

    valid = result_df[
        result_df["prediction"].isin([0, 1])
    ]

    errors = valid[
        valid["label"].astype(int)
        != valid["prediction"].astype(int)
    ].copy()

    errors["setting"] = name
    all_errors.append(errors)

    invalid = result_df[
        ~result_df["prediction"].isin([0, 1])
    ].copy()

    invalid["setting"] = name
    all_invalid.append(invalid)

errors_df = (
    pd.concat(all_errors, ignore_index=True)
    if all_errors else pd.DataFrame()
)

invalid_df = (
    pd.concat(all_invalid, ignore_index=True)
    if all_invalid else pd.DataFrame()
)

errors_df.to_csv(
    OUT_DIR / "all_errors.csv",
    index=False,
    encoding="utf-8-sig",
)

invalid_df.to_csv(
    OUT_DIR / "invalid_outputs.csv",
    index=False,
    encoding="utf-8-sig",
)

print("Classification errors:", len(errors_df))
print("Invalid-format outputs:", len(invalid_df))

if len(errors_df):
    display(
        errors_df[
            [
                "setting",
                "row_id",
                "label",
                "prediction",
                "text",
                "raw_output",
            ]
        ].head(20)
    )

if len(invalid_df):
    display(
        invalid_df[
            [
                "setting",
                "row_id",
                "label",
                "raw_output",
                "error",
            ]
        ].head(20)
    )
````
